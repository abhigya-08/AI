Problem - 4: Matrix Operations:

1. Prove

```python
import numpy as np
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

# 1
A_inv = np.linalg.inv(A)
print("1.\nA.A':\n", np.round(np.dot(A, A_inv)))

# 2
AB = np.dot(A, B)
BA = np.dot(B, A)
print("\n\n2.\nA.B:\n", AB,"\nB.A:\n", BA)

# 3
AB_T = np.transpose(AB)
A_T = np.transpose(A)
B_T = np.transpose(B)
BT_AT = np.dot(B_T, A_T)
print("\n\n3.\n(AB)T:\n", AB_T, "\n(B)T.(A)T:\n", BT_AT)
```

Solve

```python
import numpy as np

A = np.array([[2, -3, 1], [1, -1, 2], [3, 1, -1]])
B = np.array([[-1], [-3], [9]])

# A.X = B => X = A_inv.B

A_inv = np.linalg.inv(A)
X = np.dot(A_inv, B)
print("Solution:\n", X)
```

4.2 Experiment: How Fast is Numpy?

1. Element-wise Addition:

```python
import numpy as np
from timeit import timeit

#from python list
A = list(range(1000000))
B = list(range(1000000))
def add(A, B):
  return [a+b for a,b in zip(A,B)]
python_time = timeit("add(A,B)", globals=globals(), number=1000)
print(f"Time taken by python list is {python_time}s")

# from numpy arrays
A = np.arange(1000000)
B = np.arange(1000000)
def add(A, B):
  return np.add(A, B)
numpy_time = timeit("add(A,B)", globals=globals(), number=1000)
print(f"Time taken by numpy array is {numpy_time}s")
```

2. Element-wise Multiplication

```python
import numpy as np
from timeit import timeit

#from python list
A = list(range(1000000))
B = list(range(1000000))
def multiply(A, B):
  return [a*b for a,b in zip(A,B)]
python_time = timeit("multiply(A,B)", globals=globals(), number=1000)
print(f"Time taken by python list is {python_time}s")

# from numpy arrays
A = np.arange(1000000)
B = np.arange(1000000)
def multiply(A, B):
  return np.multiply(A, B)
numpy_time = timeit("multiply(A,B)", globals=globals(), number=1000)
print(f"Time taken by numpy array is {numpy_time}s")
```

3. Dot Product

```python
import numpy as np
from timeit import timeit

#from python list
A = list(range(1000000))
B = list(range(1000000))
def dot(A, B):
  return sum(a * b for a, b in zip(A, B))
python_time = timeit("dot(A,B)", globals=globals(), number=1000)
print(f"Time taken by python list is {python_time}s")

# from numpy arrays
A = np.arange(1000000)
B = np.arange(1000000)
def dot(A, B):
  return np.dot(A, B)
numpy_time = timeit("dot(A,B)", globals=globals(), number=1000)
print(f"Time taken by numpy array is {numpy_time}s")
```

4. Matrix Multiplication

```python
import numpy as np
from timeit import timeit

#from python list
A = [[1 for j in range(1000)] for i in range (1000)]
B = [[2 for j in range(1000)] for i in range (1000)]
def matrix_multiply(A, B):
  result = [[0 for j in range(1000)] for i in range(1000)]
  for i in range(1000):
    for j in range(1000):
      for k in range(1000):
        result[i][j] += A[i][k]*B[k][j]
  return result
python_time = timeit("matrix_multiply(A,B)", globals=globals(), number=1000)
print(f"Time taken by python list is {python_time}s")

# from numpy arrays
A = np.ones((1000,1000))
B = np.full((1000,1000), 2)
def matrix_multiply(A, B):
  return np.dot(A, B)
numpy_time = timeit("matrix_multiply(A,B)", globals=globals(), number=1000)
print(f"Time taken by numpy array is {numpy_time}s")
```
